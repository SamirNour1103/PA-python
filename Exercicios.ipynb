##Cálculo de Média

Um fluxograma pode representar o processo de cálculo da média de duas notas, mostrando as etapas de entrada das notas, a operação de soma e a divisão por 2 para obter a média.

image.png

```python
#entrada
num1=float(input('Digite o valor do numero 1: '))
num2=float(input('Digite o valor do numero 2: '))

#processamento
media=(num1+num2)/2

#saida
print(f"A média é: {media}")
```

##Conversão de Moeda
Este exemplo demonstra como o fluxograma de conversão de moeda pode ser implementado em Python utilizando a entrada do usuário, a cotação do dólar e a operação de divisão.

image.png

```python
cotacaomoeda=float(input('Digite o valor da cotacao do dolar: '))
valorreais=float(input('Digite o valor: '))

#processamento
resultado = valorreais/cotacaomoeda

#saida
print(f"O valor do dinheiro em dolar é: {resultado:.2f}")
```

##Cálculo de Comissão
Outro exemplo é o cálculo de comissão. O fluxograma é convertido em código Python, utilizando o preço unitário, a quantidade vendida e o cálculo da comissão.

image.png

```python
#entrada
valor=float(input('Digite o valor do item: '))
quantidade=float(input('Digite a quantidades de itens vendidos: '))

#processamento
venda=valor*quantidade
comissao=(venda*5)/100

#saida
print(f"O valor da venda foi: {venda}")
print(f"O valor da comissao foi de porcentagem: {comissao}")
```

##Calculo de hora extra
Um fluxograma pode representar o processo de cálculo de horas a mais a ser recebida, mostrando as etapas de entrada das variáveis, a operação a ser feita com divisão e soma para obter o valor final.

image.png

```python
#entrada
horas = float(input('Me diga o tanto de horas trabalhadas: '))
minutos = float(input('Me diga o tanto de minutos trabalhadas: '))
salarioporhora = float(input('Me diga o tanto que voce ganha por hora trabalhadas: '))

#processamento
conversao = minutos/60 + horas
salarioextra = (conversao*salarioporhora)*70/100 + conversao

#saida
print(f"O valor do salario extra: {salarioextra:.2f}")
```

##Investimento em CDB
 investir um valor em um CDB que rende uma taxa fixa de X% ao ano. Escreva um algoritmo que solicite o valor inicial do investimento, a taxa de juros anual (em %) e o período em anos. Depois, calcule o montante final considerando juros compostos.

image.png

```python
#entrada
valorinvestimento = float(input('Me diga o valor do investimento: '))
taxadejuros = float(input('Me diga o juros anual: '))
tempo = float(input('Me diga o tempo em anos : '))

#processamento
jurosdec = taxadejuros/100
montante = valorinvestimento*(1+jurosdec)**tempo

#saida
print(f'O valor do investimento após o tempo é de: {montante:.2f}')
```

##Divisão de Conta em um Restaurante com Gorjeta
Dividir a conta igualmente entre amigos. Além disso, o restaurante cobra uma taxa de serviço de 10% sobre o valor total da conta mais uma gorjeta.

image.png

```python
#entrada
valorconta = float(input('Me diga o valor da conta: '))
numpessoas = float(input('Qual o numero de pessoas: '))
gorjeta =  float(input('O valor da gorjeta é qual: '))
taxa = 10/100

#processamento
taxaservi = (taxa*valorconta)
valortotal = (taxaservi + valorconta) + gorjeta
valorpessoas = valortotal/numpessoas

#saida
print(f"O valor por pessoas: {valorpessoas:.2f}")
```
