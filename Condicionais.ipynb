##Calcule a média percentual de faltas

```python
#entrada
nota1 = float(input("Digite sua primeira nota: "))
nota2 = float(input("Digite sua segunda nota: "))
nota3 = float(input("Digite sua terceira nota: "))
dias = 200
faltas = int(input("Digite suas faltas: "))


#processamento
media = (nota1 + nota2 + nota3) /3
precenca = 100-((faltas / dias)* 100)

if media>=6 and precenca>=75:
  print(f'Você foi aprovado, com a media de {media:.2f}' ' e com o percentual de precença de',precenca,'%')
else:
   print(f'Você foi reprovado, com a media de {media:.2f}' ' e com o percentual de precença de',precenca,'%')
```

```python
if (l1 + l2 > l3) and (l1 + l3 > l2) and (l2 + l3 > l1):
  if (l1 == l2) and (l2 == l3):
    print("Esses valores formam um triângulo equilátero")
  elif (l1 != l2) and (l2 != l3) and (l1 != l3):
    print("Esses valores formam um triângulo escaleno")
  elif (l1 == l2) and (l2 != l3) or (l1 == l3) and (l2 != l3)  or (l2 == l3) and (l1 != l3):
    print("Esses valores formam um triângulo isósceles")
else:
  print("Esses valores não formam um triângulo")
```

#10 - Simulador de frete inteligente

Peça o peso do pacote (em kg) e a distância (em km) de sua cidade até o local de entrega.Calcule o frete com base em regras:

Até 2kg → R$5 por 100km

De 2kg a 5kg → R$8 por 100km

Acima de 5kg → R$10 por 100km

Exemplo de saída:

"Para um pacote de 3kg viajando 350km, o valor do frete será R$28,00"

```python
peso = float(input("Informe o peso do pacote (em kg): "))
distancia = float(input("Informe a distância (em km): "))
```

```python
if peso <= 2:
    valorkm = 5
elif (peso > 2) or (peso <= 5):
    valorkm = 8
else:
    valorkm = 10

frete = (distancia / 100) * valorkm

print(f"Para um pacote de {peso}kg viajando {distancia}km, o valor do frete será R${frete:.2f}")
```
