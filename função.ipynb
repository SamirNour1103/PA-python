Crie uma função chamada gerar_senha() que não recebe parâmetros.

Ela retorna uma senha aleatória de 8 caracteres contendo letras, números e caracteres especiais;

Essa senha deve ter letra maiúscula, minúscula, um número e um caractere especial.

```python
import random
import string

maiscula = random.choice(string.ascii_uppercase)
```

```python
def gerarsenha():
  senha = ''
  for i in range(2):
    senha = senha + random.choice(string.ascii_uppercase) + random.choice(string.punctuation) + random.choice(string.ascii_lowercase) + random.choice(string.digits)
  return senha
```

#2) Crie uma função com parÂmetro e sem retorno
 chamada mostrar_tabuada(numero) que recebe um parâmetro inteiro e imprime na tela a tabuada de 1 a 10 desse número.

```python
def mostrar_tabuada(numero):
  numero = int(input('Me diga um numero: '))
  tabuada = ""
  for i in range(11):
    tabuada = (numero * i)
    print(f'{numero} X {i} = {tabuada}')
```

```python
mostrar_tabuada(5)
```

##3) Crie uma função com retorno e com parâmetro chamada validar_senha(senha) que recebe uma string e retorna um dicionário contendo:

Chave Significado

tamanho_valido

Verdadeiro se tiver 8 caracteres

tem_numero

Verdadeiro se possuir nümero

tem maiuscula

Verdadeiro se tiver letra maiúscula

tem_minuscula

Verdadeiro se tiver letra minúscula

tem_especial

Verdadeiro se possuir caractere especial

senha_valida

Verdadeiro apenas se todos os critérios

forem atendidos

```python
def validar_senha(senha):
    resultado = {
        "tamanho_valido": len(senha) >= 8,
        "tem_numero": False,
        "tem_maiuscula": False,
        "tem_minuscula": False,
        "tem_especial": False,
        "senha_valida": False,
        "mensagem": ""
    }


    # senha é válida apenas se todos os critérios forem atendidos
    resultado["senha_valida"] = (
        resultado["tamanho_valido"] and
        resultado["tem_numero"] and
        resultado["tem_maiuscula"] and
        resultado["tem_minuscula"] and
        resultado["tem_especial"]
    )

    # mensagem final
    print("Todos os critérios foram atendidos. A senha é válida.")

    return resultado
```

```python
print(validar_senha(gerarsenha()))
```
