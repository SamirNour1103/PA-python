##Exercício 5: Analisando uma Lista de Tuplas

- Crie uma lista de tuplas, onde cada tupla contém um par (nome, idade).
- Encontre e imprima o nome da pessoa mais velha.
- Encontre e imprima a idade média das pessoas.

```python
# seu código aqui...


pessoas = [("João", 125), ("Maria", 30), ("Pedro", 35), ("Ana", 20), ("Carlos", 28)]
```

```python
if pessoas[0][1] > pessoas[1][1]:
    nomevelho = pessoas[0][0]
    idade = pessoas[0][1]
else:
    nomevelho = pessoas[1][0]
    idade = pessoas[1][1]

if pessoas[2][1] > idade:
    nomevelho = pessoas[2][0]
    idade = pessoas[2][1]

if pessoas[3][1] > idade:
    nomevelho = pessoas[3][0]
    idade = pessoas[3][1]

if pessoas[4][1] > idade:
    nomevelho = pessoas[4][0]
    idade = pessoas[4][1]

total_idades = pessoas[0][1] + pessoas[1][1] + pessoas[2][1] + pessoas[3][1] + pessoas[4][1]
quantidade_pessoas = len(pessoas)
idade_media = total_idades / quantidade_pessoas

print(f"A pessoa mais velha é {nomevelho} com {idade} anos.")
print(f"A idade média das pessoas é {idade_media:.2f} anos.")
```

```python
# 8 - Faça um programa que retorne True se duas strings lidas do teclado são palíndromas mútuas, ou seja, se uma é igual à outra quando lida de traz para frente ou False, se não forem.
# Por exemplo: amor e roma

frase1 = input('Media uma frase: ')
frase2 = input('Media uma frase: ')

frase2 == frase2 [::-1]

if frase1[::-1] == frase2:
  print('A palavra é um palindromo')
elif frase1[::-1] == frase1:
  print('A palavra 1 é palindromo com ela mesma')
elif frase2[::-1] == frase2:
  print('A palavra 2 é palindromo com ela mesma')
else:
   print('Não é palindromo')
```
